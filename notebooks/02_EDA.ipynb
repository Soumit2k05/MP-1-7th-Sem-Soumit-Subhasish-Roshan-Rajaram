**Import Libraries**

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

**Load Cleaned Dataset**

```python
df = pd.read_csv("/content/Energy_consumption_dataset.csv")

print("Dataset Loaded Successfully")
```

**View Dataset**

```python
print(df.head())

print(df.info())

print(df.describe())
```

**Time Series Plot**

```python
plt.figure(figsize=(12,5))
plt.plot(df["EnergyConsumption"])
plt.title("Energy Consumption Over Time")
plt.xlabel("Index")
plt.ylabel("Power")
plt.show()
```

**Histogram**

```python
df["EnergyConsumption"].hist(figsize=(8,5))
plt.title("Histogram of Energy Consumption")
plt.show()
```

**Monthly Trend**

```python
# The 'Month' column already exists in your dataset
monthly = df.groupby("Month")["EnergyConsumption"].mean()

monthly.plot(kind="line", marker="o", figsize=(10,5))
plt.title("Monthly Average Power Consumption")
plt.xlabel("Month")
plt.ylabel("Average Power")
plt.show()
```

**Weekly Trend**

```python
# Since there is no 'Datetime' column, we skip Week for now or base it on available data
# Using DayOfWeek instead
weekly = df.groupby("DayOfWeek")["EnergyConsumption"].mean()

weekly.plot(kind="bar", figsize=(10,5))
plt.title("Average Power Consumption by Day of Week")
plt.ylabel("Average Power")
plt.show()
```

**Hour-wise Trend**

```python
# The 'Hour' column already exists in your dataset
hourly = df.groupby("Hour")["EnergyConsumption"].mean()

hourly.plot(kind="line", marker="o", figsize=(10,5))
plt.title("Hour-wise Average Power Consumption")
plt.xlabel("Hour")
plt.ylabel("Average Power")
plt.show()
```
