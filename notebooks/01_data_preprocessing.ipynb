## Step 7 - Identify Data Quality Issues
Check for missing values (since sensor readings can drop out), duplicate rows (which can bias the model), incorrect data types (numeric columns stored as text must be converted), and outliers (unusually high or low readings that may indicate sensor errors). Each of these, if left untreated, can reduce model accuracy.

```python
print("--- Missing Values ---")
print(df.isnull().sum())

print(f"\n--- Duplicate Rows ---")
print(df.duplicated().sum())
```
